```python
import pandas as pd
import matplotlib.pyplot as plt
from mwclient import Site
from mwviews.api import PageviewsClient
```

```python
kdramas = pd.read_csv("/Users/wuyuwei/Desktop/kdramas_2023.csv")
```

```python
kdramas.head()
```

```python
site = Site("en.wikipedia.org")
```

```python
dramas = [
    "The Glory (TV series)",
    "Moving (South Korean TV series)",
    "My Demon",
    "King the Land",
    "Gyeongseong Creature",
    "Twinkling Watermelon",
    "Death's Game",
    "Crash Course in Romance",
    "Welcome to Samdal-ri",
    "Celebrity (South Korean TV series)",
    "A Time Called You",
    "Perfect Marriage Revenge",
    "See You in My 19th Life",
    "The Good Bad Mother",
    "My Lovely Liar",
    "Doona!",
    "Love to Hate You (TV series)",
    "Daily Dose of Sunshine",
    "Destined with You"
]
```

```python
for drama in dramas:
    page = site.pages[drama]
    print(drama, ", page: ", page)
```

```python
p = PageviewsClient(user_agent="tw106@wellesley.edu")
```

```python
data = p.article_views(
    "en.wikipedia",
    dramas,
    start="20220601",
    end="20240601"
)
```

```python
print(type(data))
```

```python
list(data.items())[:3]
```

```python
pageviews_df = pd.DataFrame(data)
pageviews_df.head()
```

```python
pageviews_df.columns = pd.to_datetime(pageviews_df.columns)
```

```python
total_pageviews = pageviews_df.sum(axis=1)
total_pageviews
```

```python
total_df = total_pageviews.reset_index()
total_df.columns = ["Drama", "Total Pageviews"]
total_df
```

```python
total_df = total_df.sort_values(
    "Total Pageviews",
    ascending=True
)
total_df
```

```python
plt.figure(figsize=(12, 10))
plt.barh(
    total_df["Drama"],
    total_df["Total Pageviews"]
)
plt.xlabel("Total Pageviews")
plt.xlabel("Total Pageviews")
plt.title("Pageviews per Show (2022-06-01 to 2024-06-01)")
plt.show()
```

## Graph 2

```python
peak_dates = pageviews_df.idxmax(axis=1)
peak_dates
```

```python
peak_values = pageviews_df.max(axis=1)
peak_values
```

```python
drama = 'The_Glory_(TV_series)'
pageviews_df.loc[drama]
```

```python
peak_date = pageviews_df.loc[drama].idxmax()
peak_date
```

```python
days_relative_to_peak = (
    pageviews_df.columns - peak_date
).days

days_relative_to_peak[:10]
```

```python
aligned_glory = pd.DataFrame({
    "Days": days_relative_to_peak,
    "Pageviews": pageviews_df.loc[drama].values  # 改用 .loc
})

aligned_glory.head()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    aligned_glory["Days"],
    aligned_glory["Pageviews"]
)

plt.fill_between(
    aligned_glory["Days"],
    aligned_glory["Pageviews"]
)

plt.xlabel("Days Relative to Peak Date (0)")
plt.ylabel("Pageviews")

plt.title(
    "The Glory — Pageviews Aligned to Peak Date"
)

plt.show()

```

```python
aligned_data = {}
```

```python
for drama in pageviews_df.index:
    
    peak_date = pageviews_df.loc[drama].idxmax()
    
    days_relative_to_peak = (
        pageviews_df.columns - peak_date
    ).days
    
    aligned_data[drama] = {
        "Days": days_relative_to_peak,
        "Pageviews": pageviews_df.loc[drama].values
    }
```

```python
list(aligned_data.keys())
```

```python
drama_list = list(aligned_data.keys())

fig, axes = plt.subplots(len(drama_list), 1, figsize=(10, 15), sharex=True)
for i, drama in enumerate(drama_list):
    ax = axes[i]
    
    days = aligned_data[drama]["Days"]
    pageviews = aligned_data[drama]["Pageviews"]
    ax.plot(days, pageviews)
    ax.fill_between(days, pageviews, alpha=0.3)
    ax.axvline(0, color='red', linestyle='--')
    ax.set_ylabel(f"[{i+1}] {drama}", rotation=0, ha='right', va='center')

axes[-1].set_xlabel("Days Relative to Peak Date (0)")
fig.suptitle("Aligned Pageview Trajectories (Day 0 = Peak Date)")
plt.xlim(-120, 120)

plt.tight_layout()
plt.show()
```
